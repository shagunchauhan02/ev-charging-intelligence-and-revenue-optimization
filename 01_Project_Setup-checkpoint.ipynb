```python
import pandas as pd
import numpy as np

# Reproducibility
np.random.seed(42)

# Number of charging sessions
n = 50000

# Session IDs
session_ids = [f"EVS{i:05d}" for i in range(1, n + 1)]

# Station information
station_ids = np.random.choice(
    [f"ST{i:03d}" for i in range(1, 31)],
    size=n
)

station_names = {
    "ST001": "Delhi Central",
    "ST002": "Noida Sector 18",
    "ST003": "Gurgaon Cyber Hub",
    "ST004": "Meerut Bypass",
    "ST005": "Ghaziabad Highway",
    "ST006": "Jaipur Central",
    "ST007": "Lucknow Expressway",
    "ST008": "Agra Highway",
    "ST009": "Chandigarh Sector 17",
    "ST010": "Dehradun Central",
    "ST011": "Haridwar Highway",
    "ST012": "Kanpur Central",
    "ST013": "Faridabad Sector 15",
    "ST014": "Delhi Airport",
    "ST015": "Noida Expressway",
    "ST016": "Gurgaon Sohna Road",
    "ST017": "Agra City",
    "ST018": "Jaipur Highway",
    "ST019": "Lucknow City",
    "ST020": "Chandigarh Highway",
    "ST021": "Delhi Rohini",
    "ST022": "Ghaziabad City",
    "ST023": "Meerut City",
    "ST024": "Dehradun Highway",
    "ST025": "Haridwar City",
    "ST026": "Kanpur Highway",
    "ST027": "Faridabad Highway",
    "ST028": "Delhi Dwarka",
    "ST029": "Noida Sector 62",
    "ST030": "Gurgaon Golf Course"
}

cities = {
    "ST001": "Delhi",
    "ST002": "Noida",
    "ST003": "Gurgaon",
    "ST004": "Meerut",
    "ST005": "Ghaziabad",
    "ST006": "Jaipur",
    "ST007": "Lucknow",
    "ST008": "Agra",
    "ST009": "Chandigarh",
    "ST010": "Dehradun",
    "ST011": "Haridwar",
    "ST012": "Kanpur",
    "ST013": "Faridabad",
    "ST014": "Delhi",
    "ST015": "Noida",
    "ST016": "Gurgaon",
    "ST017": "Agra",
    "ST018": "Jaipur",
    "ST019": "Lucknow",
    "ST020": "Chandigarh",
    "ST021": "Delhi",
    "ST022": "Ghaziabad",
    "ST023": "Meerut",
    "ST024": "Dehradun",
    "ST025": "Haridwar",
    "ST026": "Kanpur",
    "ST027": "Faridabad",
    "ST028": "Delhi",
    "ST029": "Noida",
    "ST030": "Gurgaon"
}

charger_type = np.random.choice(
    ["AC", "DC Fast"],
    size=n,
    p=[0.40, 0.60]
)

connector_type = np.random.choice(
    ["CCS2", "Type 2", "CHAdeMO"],
    size=n,
    p=[0.65, 0.30, 0.05]
)

vehicle_type = np.random.choice(
    ["Sedan", "SUV", "Hatchback"],
    size=n,
    p=[0.40, 0.40, 0.20]
)

customer_type = np.random.choice(
    ["Member", "Guest"],
    size=n,
    p=[0.60, 0.40]
)

payment_method = np.random.choice(
    ["UPI", "Credit Card", "Debit Card", "Wallet"],
    size=n,
    p=[0.55, 0.20, 0.15, 0.10]
)

battery_capacity = np.random.choice(
    [40, 50, 60, 75, 82, 100],
    size=n,
    p=[0.15, 0.15, 0.25, 0.20, 0.15, 0.10]
)

# Start dates across one year
start_time = pd.to_datetime(
    np.random.uniform(
        pd.Timestamp("2025-01-01").value,
        pd.Timestamp("2025-12-31 23:59:59").value,
        n
    )
)

# Charging duration in minutes
charging_duration = np.random.randint(15, 181, size=n)

# End time
end_time = start_time + pd.to_timedelta(
    charging_duration,
    unit="m"
)

# Energy consumed
energy_kwh = np.round(
    np.random.uniform(8, 80, size=n),
    2
)

# Price per kWh
price_per_kwh = np.random.choice(
    [12, 15, 18, 20, 22],
    size=n,
    p=[0.10, 0.25, 0.35, 0.20, 0.10]
)

# Revenue
revenue = np.round(
    energy_kwh * price_per_kwh,
    2
)

# Station status
station_status = np.random.choice(
    ["Active", "Maintenance"],
    size=n,
    p=[0.95, 0.05]
)

# Create dataframe
df = pd.DataFrame({
    "Session_ID": session_ids,
    "Station_ID": station_ids,
    "Station_Name": [station_names[x] for x in station_ids],
    "City": [cities[x] for x in station_ids],
    "Charger_Type": charger_type,
    "Connector_Type": connector_type,
    "Start_Time": start_time,
    "End_Time": end_time,
    "Charging_Duration_Min": charging_duration,
    "Energy_kWh": energy_kwh,
    "Price_per_kWh": price_per_kwh,
    "Revenue": revenue,
    "Vehicle_Type": vehicle_type,
    "Battery_Capacity_kWh": battery_capacity,
    "Customer_Type": customer_type,
    "Payment_Method": payment_method,
    "Station_Status": station_status
})

df.head()
```
