```python
df["Hour"] = df["Start_Time"].dt.hour

df["Month_Number"] = df["Start_Time"].dt.month

df["Month_Name"] = df["Start_Time"].dt.month_name()

df["Day_of_Week"] = df["Start_Time"].dt.day_name()

df["Day_Number"] = df["Start_Time"].dt.dayofweek
```

```python
def classify_period(hour):
    if 7 <= hour <= 10 or 17 <= hour <= 22:
        return "Peak"
    else:
        return "Off-Peak"

df["Demand_Period"] = df["Hour"].apply(classify_period)
```

```python
df["Day_Type"] = np.where(
    df["Day_Number"] >= 5,
    "Weekend",
    "Weekday"
)
```

```python
station_metrics = (
    df.groupby(["Station_ID", "Station_Name", "City"])
    .agg(
        Total_Sessions=("Session_ID", "nunique"),
        Total_Energy_kWh=("Energy_kWh", "sum"),
        Total_Revenue=("Revenue", "sum"),
        Avg_Energy_kWh=("Energy_kWh", "mean"),
        Avg_Revenue=("Revenue", "mean"),
        Avg_Duration_Min=("Charging_Duration_Min", "mean")
    )
    .reset_index()
)
```

```python
station_metrics.head()
```

```python
station_metrics["Revenue_per_Session"] = (
    station_metrics["Total_Revenue"] /
    station_metrics["Total_Sessions"]
).round(2)
```

```python
station_metrics["Energy_per_Session"] = (
    station_metrics["Total_Energy_kWh"] /
    station_metrics["Total_Sessions"]
).round(2)
```

```python
station_metrics["Revenue_Rank"] = (
    station_metrics["Total_Revenue"]
    .rank(method="dense", ascending=False)
    .astype(int)
)
```

```python
station_metrics.sort_values(
    "Revenue_Rank"
).head(10)
```

```python
station_metrics.info()
```

```python
station_metrics["Revenue_Score"] = (
    station_metrics["Total_Revenue"] /
    station_metrics["Total_Revenue"].max()
)

station_metrics["Session_Score"] = (
    station_metrics["Total_Sessions"] /
    station_metrics["Total_Sessions"].max()
)

station_metrics["Energy_Score"] = (
    station_metrics["Total_Energy_kWh"] /
    station_metrics["Total_Energy_kWh"].max()
)
```

```python
station_metrics.head()
```

```python
station_metrics["Performance_Score"] = (
    station_metrics["Revenue_Score"] * 0.5 +
    station_metrics["Session_Score"] * 0.3 +
    station_metrics["Energy_Score"] * 0.2
).round(3)
```

```python
station_metrics[
    [
        "Station_Name",
        "Total_Revenue",
        "Total_Sessions",
        "Performance_Score"
    ]
].head(10)
```

```python
def performance_category(score):
    if score >= 0.75:
        return "Excellent"
    elif score >= 0.50:
        return "Good"
    elif score >= 0.25:
        return "Average"
    else:
        return "Underperforming"

station_metrics["Performance_Category"] = (
    station_metrics["Performance_Score"]
    .apply(performance_category)
)
```

```python
station_metrics["Performance_Category"].value_counts()
```

```python
performance_counts = (
    station_metrics["Performance_Category"]
    .value_counts()
)

plt.figure(figsize=(8, 5))

performance_counts.plot(kind="bar")

plt.title("Charging Station Performance Categories")
plt.xlabel("Performance Category")
plt.ylabel("Number of Stations")

plt.tight_layout()
plt.show()
```

```python
revenue_threshold = station_metrics["Total_Revenue"].quantile(0.75)
session_threshold = station_metrics["Total_Sessions"].quantile(0.75)

print("Revenue Threshold:", revenue_threshold)
print("Session Threshold:", session_threshold)
```

```python
expansion_candidates = station_metrics[
    (station_metrics["Total_Revenue"] >= revenue_threshold) &
    (station_metrics["Total_Sessions"] >= session_threshold)
].sort_values(
    "Performance_Score",
    ascending=False
)
```

```python
# Check required columns first

required_columns = [
    "Utilization_Rate",
    "Revenue_per_Charger",
    "Total_Sessions"
]

print(station_metrics[required_columns].head())
```

```python
# Create scores manually using min-max normalization

station_metrics["Utilization_Score"] = (
    (station_metrics["Utilization_Rate"] -
     station_metrics["Utilization_Rate"].min()) /
    (station_metrics["Utilization_Rate"].max() -
     station_metrics["Utilization_Rate"].min())
)

station_metrics["Revenue_Efficiency_Score"] = (
    (station_metrics["Revenue_per_Charger"] -
     station_metrics["Revenue_per_Charger"].min()) /
    (station_metrics["Revenue_per_Charger"].max() -
     station_metrics["Revenue_per_Charger"].min())
)

station_metrics["Demand_Score"] = (
    (station_metrics["Total_Sessions"] -
     station_metrics["Total_Sessions"].min()) /
    (station_metrics["Total_Sessions"].max() -
     station_metrics["Total_Sessions"].min())
)

print("Scores created successfully!")
```

```python
station_metrics[
    [
        "Station_Name",
        "Utilization_Score",
        "Revenue_Efficiency_Score",
        "Demand_Score"
    ]
].head(10)
```

```python
station_metrics["Expansion_Score"] = (
    station_metrics["Utilization_Score"] * 0.45 +
    station_metrics["Revenue_Efficiency_Score"] * 0.30 +
    station_metrics["Demand_Score"] * 0.25
).round(3)

print("Expansion score created successfully!")
```

```python
station_metrics[
    [
        "Station_Name",
        "Utilization_Percent",
        "Revenue_per_Charger",
        "Expansion_Score"
    ]
].sort_values(
    "Expansion_Score",
    ascending=False
).head(10)
```

```python
def expansion_priority(score):
    if score >= 0.75:
        return "High Priority"
    elif score >= 0.50:
        return "Medium Priority"
    else:
        return "Low Priority"


station_metrics["Expansion_Priority"] = (
    station_metrics["Expansion_Score"]
    .apply(expansion_priority)
)
```
